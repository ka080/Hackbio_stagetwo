```python
def dna_to_protein(dna_seq):
    """Translate a DNA sequence into a protein sequence."""
    codon_table = {
        'ATA':'I', 'ATC':'I', 'ATT':'I', 'ATG':'M',
        'ACA':'T', 'ACC':'T', 'ACG':'T', 'ACT':'T',
        'AAC':'N', 'AAT':'N', 'AAA':'K', 'AAG':'K',
        'AGC':'S', 'AGT':'S', 'AGA':'R', 'AGG':'R',
        'CTA':'L', 'CTC':'L', 'CTG':'L', 'CTT':'L',
        'CCA':'P', 'CCC':'P', 'CCG':'P', 'CCT':'P',
        'CAC':'H', 'CAT':'H', 'CAA':'Q', 'CAG':'Q',
        'CGA':'R', 'CGC':'R', 'CGG':'R', 'CGT':'R',
        'GTA':'V', 'GTC':'V', 'GTG':'V', 'GTT':'V',
        'GCA':'A', 'GCC':'A', 'GCG':'A', 'GCT':'A',
        'GAC':'D', 'GAT':'D', 'GAA':'E', 'GAG':'E',
        'GGA':'G', 'GGC':'G', 'GGG':'G', 'GGT':'G',
        'TCA':'S', 'TCC':'S', 'TCG':'S', 'TCT':'S',
        'TTC':'F', 'TTT':'F', 'TTA':'L', 'TTG':'L',
        'TAC':'Y', 'TAT':'Y', 'TAA':'_', 'TAG':'_', 'TGA':'_'
    }

    protein = ""
    # Process dna in three nucleotides
    for i in range(0, len(dna_seq) - 2, 3):
        codon = dna_seq[i:i+3].upper()
        protein += codon_table.get(codon, 'X')  # 'X' for unknown codons
    return protein

# Example
dna_sequence = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"
print(dna_to_protein(dna_sequence))
```

```python
def hamming_distance(str1, str2):
    """Calculate the Hamming distance between two strings."""
    #  shorter string with underscores (_) to equalize length
    max_len = max(len(str1), len(str2))
    str1 = str1.ljust(max_len, '_')
    str2 = str2.ljust(max_len, '_')

    # Count differing characters
    return sum(ch1 != ch2 for ch1, ch2 in zip(str1, str2))


slack_username = "kashish"
twitter_handle = "kranzkafka"

distance = hamming_distance(slack_username, twitter_handle)
print(f"Hamming distance between '{slack_username}' and '{twitter_handle}' is: {distance}")
```
